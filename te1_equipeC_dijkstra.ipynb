## Partie A --- La matrice pondérée

Comme à la séance 4 : la case ne vaut plus 0 ou 1, elle vaut **le nombre de minutes**.
Un 0 veut dire : **pas de rue**.

La matrice est donnée. Vérifiez seulement qu'elle est symétrique.

```python
sommets = ["A", "B", "C", "D", "E", "F", "G"]

#           A  B  C  D  E  F  G
matrice = [[0, 2, 5, 0, 0, 0, 0],   # A
           [2, 0, 1, 4, 0, 0, 0],   # B
           [5, 1, 0, 6, 3, 0, 0],   # C
           [0, 4, 6, 0, 2, 0, 7],   # D
           [0, 0, 3, 2, 0, 1, 0],   # E
           [0, 0, 0, 0, 1, 0, 3],   # F
           [0, 0, 0, 7, 0, 3, 0]]   # G


def est_symetrique(matrice):
    for i in range(len(matrice)):
        for j in range(len(matrice)):
            if matrice[i][j] != matrice[j][i]:
                return False
    return True

print("symetrique :", est_symetrique(matrice))
```

## Partie B --- Le poids d'un chemin

Rappel : **le poids d'un chemin est la somme des nombres écrits sur les rues qu'il emprunte.**
Un chemin de 4 quartiers contient 3 rues : c'est le `- 1` de la boucle.

La fonction `numero` est donnée. Complétez `poids`.

```python
def numero(sommets, nom):
    """Le numero de ligne du quartier nom."""
    for i in range(len(sommets)):
        if sommets[i] == nom:
            return i
    return -1


def poids(matrice, sommets, chemin):
    """La somme des minutes le long du chemin."""
    total = 0
    for k in range(len(chemin) - 1):
        i = numero(sommets, chemin[k])
        j = numero(sommets, chemin[k + 1])
        # TODO 1 : ajouter au total les minutes de la rue entre i et j
    return total


print(poids(matrice, sommets, ["A", "B", "D", "G"]))
print(poids(matrice, sommets, ["A", "B", "C", "E", "F", "G"]))
```

**Vous devez obtenir `13` puis `10`.**

Remarquez : le premier chemin fait **3 rues**, le second en fait **5**.
Et pourtant c'est le second qui est le plus rapide.

## Partie F --- Et le BFS, il aurait dit quoi ?

Le BFS ne voit **pas** les minutes : il compte les rues. Le code est donné, faites-le tourner.

```python
from collections import deque

# le meme reseau, mais sans les poids : un dictionnaire de voisins
reseau = {}
for i in range(len(sommets)):
    voisins = []
    for j in range(len(sommets)):
        if matrice[i][j] > 0:
            voisins.append(sommets[j])
    reseau[sommets[i]] = voisins

def niveaux(graphe, depart):
    niveau = {depart: 0}
    file = deque([depart])
    while file:
        sommet = file.popleft()
        for voisin in graphe[sommet]:
            if voisin not in niveau:
                niveau[voisin] = niveau[sommet] + 1
                file.append(voisin)
    return niveau

niv = niveaux(reseau, "A")
print("reseau sans les poids :", reseau)
print("niveau de G (en nombre de rues) :", niv["G"])
print("le chemin A, B, D, G fait", len(["A", "B", "D", "G"]) - 1, "rues et",
      poids(matrice, sommets, ["A", "B", "D", "G"]), "minutes")
```
